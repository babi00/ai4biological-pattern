```python
import os
import shutil

def copy_next_batch(src_folder, dst_folder, batch_size=1500):
    os.makedirs(dst_folder, exist_ok=True)

    # Only select files that are in the source but not yet in destination
    all_src_files = sorted([
        f for f in os.listdir(src_folder)
        if os.path.isfile(os.path.join(src_folder, f))
    ])
    already_copied = set(os.listdir(dst_folder))
    to_copy = [f for f in all_src_files if f not in already_copied]

    if not to_copy:
        print("✅ All files already copied.")
        return

    batch = to_copy[:batch_size]
    print(f"📁 Copying next {len(batch)} files...")

    for file_name in batch:
        src_path = os.path.join(src_folder, file_name)
        dst_path = os.path.join(dst_folder, file_name)
        shutil.copy2(src_path, dst_path)

    print(f"✅ Copied {len(batch)} files. Now add/commit/push before running again.")

# Example usage
source_folder = "C:\\Users\\babac\\Downloads\\taxas\\taxas\\lythrum_salicaria"
destination_folder = "lythrum_salicaria"
copy_next_batch(source_folder, destination_folder)

```

```python
import os
import pandas as pd

def clean_csvs_in_folder(folder_path, output_folder=None):
    if output_folder and not os.path.exists(output_folder):
        os.makedirs(output_folder)
    elif not output_folder:
        output_folder = folder_path  # overwrite in-place

    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            input_path = os.path.join(folder_path, filename)
            output_path = os.path.join(output_folder, filename)

            df = pd.read_csv(input_path)

            # Filter out rows where 'filename' ends with a dot
            df['filename'] = df['filename'].astype(str)
            df_cleaned = df[~df['filename'].str.endswith('.')]
            df_cleaned = df[df['filename'] != "nan"]

            # Overwrite the original CSV file with the cleaned DataFrame
            df_cleaned.to_csv(output_path, index=False)

            print(f"Cleaned CSV saved to {output_path}")
def clean_csvs_without_label(folder_path, output_folder=None):
    if output_folder and not os.path.exists(output_folder):
        os.makedirs(output_folder)
    elif not output_folder:
        output_folder = folder_path  # overwrite in-place

    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            input_path = os.path.join(folder_path, filename)
            output_path = os.path.join(output_folder, filename)

            df = pd.read_csv(input_path)

            # Filter out rows where 'filename' ends with a dot
            df = df.dropna(subset=['label'])

            # Overwrite the original CSV file with the cleaned DataFrame
            df.to_csv(output_path, index=False)

            print(f"Cleaned CSV saved to {output_path}")


rep_file="metadata"
out_folder="filtered_metadata"
clean_csvs_without_label(rep_file, out_folder)
```
