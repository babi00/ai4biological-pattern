Classify the species non based on the location, but based exclusively on the potential of a species to be invasive

```python
import os
import pandas as pd
```

```python
#manually identified invasive species
INVASIVE_SPECIES = ['lythrum hyssopifolia', 'lythrum salicaria', 'lythrum virgatum']

METADATA_PATH = './support_files/taxas/taxas/metadata'
```

```python
for dir in os.listdir(METADATA_PATH):
    species = dir.strip('.csv')
    species_split = species.split(sep='_')
    species = species_split[0] + " " + species_split[1]

    path = METADATA_PATH + '/' + dir

    if species in INVASIVE_SPECIES:
        label = 'Invasive'
    else:
        label = 'Non Invasive'

    metadata_df = pd.read_csv(path)

    metadata_df['label'] = label

    metadata_df.to_csv(path, sep=',', index=False)
```
