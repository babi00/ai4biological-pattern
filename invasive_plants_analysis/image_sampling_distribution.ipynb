Comparison of the distribution between the total number of images and the sampling executed for the clustering.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import numpy as np
import scipy
%matplotlib inline
```

```python
FILTERED_METADATA_PATH = './taxas/filtered_metadata'
INVASIVE_SPECIES = ['lythrum hyssopifolia', 'lythrum salicaria', 'lythrum virgatum']
```

```python
#DIFFERENTLY FROM WHAT HAS BEEN DONE BEFORE, WE DON'T CONSIDER HYSSOPIFOLIA IN THE SET, SINCE WE ARE NOT USING IT FOR OUR ANALYSIS

species_dict = {}

for filename in os.listdir(FILTERED_METADATA_PATH):

    species_name = filename.strip('.csv').split(sep='_')[0] + "_" + filename.strip('.csv').split(sep='_')[1]

    species_path = FILTERED_METADATA_PATH + '/' + filename
    species_df = pd.read_csv(species_path)
    
    if len(species_df!=0) and species_name != 'lythrum_hyssopifolia':
        species_dict[species_name] = len(species_df)

print(species_dict)

tot_elements = sum(species_dict.values())

print(tot_elements)

#new format 

for k,v in species_dict.items():

    species_dict[k] = {
        'items' : v,
        'percentage' : v/tot_elements
    }

species_dict = dict(sorted(species_dict.items(), key = lambda item: -item[1]['items']))

print(species_dict)
```

```python
#TAKEN FROM BARBARA
samples_dict = {'lythrum_salicaria': 741, 'lythrum_californicum': 123, 'lythrum_ovalifolium': 12, 'lythrum_alatum': 610, 'lythrum_junceum': 167, 'lythrum_portula': 126, 'lythrum_volgense': 4, 'lythrum_virgatum': 119, 'lythrum_gracile': 5, 'lythrum_lineare': 21, 'lythrum_borysthenicum': 8, 'lythrum_flexuosum': 5, 'lythrum_tribracteatum': 18, 'lythrum_acutangulum': 2, 'lythrum_flagellare': 18, 'lythrum_maritimum': 11, 'lythrum_curtissii': 1, 'lythrum_thymifolia': 5, 'lythrum_rotundifolium': 2, 'lythrum_intermedium': 1, 'lythrum_vulneraria': 1}
tot_samples = sum(samples_dict.values())
print(tot_samples)

for k,v in samples_dict.items():

    samples_dict[k] = {
        'items' : v,
        'percentage' : v/tot_samples
    }

samples_dict = dict(sorted(samples_dict.items(), key = lambda item: -item[1]['items']))
print(samples_dict)
```

```python
for k in species_dict.keys():
    if k not in samples_dict.keys():
        samples_dict[k] = {'items': 0, 'percentage': 0}
```

```python
print(len(samples_dict))
print(len(species_dict))
```

```python
species_perc = [x['percentage'] for x in list(species_dict.values())]
print(len(species_perc))
samples_perc = [x['percentage'] for x in list(samples_dict.values())]
print(len(samples_perc))

distances = [abs(t[0] - t[1]) for t in list(zip(species_perc, samples_perc))]

avg_dist = np.average(distances)
std_dist = np.std(distances)

print(f"Average distance in percentage: {avg_dist:.4f} +- {std_dist:.4f}")
```

```python
kl_dist = sum(scipy.special.rel_entr(samples_perc, species_perc)) 
#entropy = P(i) * log2(P(i)/Q(i)) where P is the observed data and Q the modelled distribution
#rel_entr returns elementwise entropy  = x*log(x/y), we want the sum of the relative entropies
print(kl_dist)

#0.002867351339360848 is very close to 0, so the distance is very low

# I. Csiszar. "
# I
# -Divergence Geometry of Probability Distributions and Minimization Problems." Ann. Probab. 3 (1) 146 - 158, February, 1975. https://doi.org/10.1214/aop/1176996454
```
